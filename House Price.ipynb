```python
import pandas as pd

# Load Dataset
df = pd.read_csv("HousePricePrediction.csv")

df.sample(10)
```

```python
# S 1 :Data Cleaning

# Rows where sales price is missing

df = df.dropna(subset=["SalePrice"])

# Handle Missing Values

df["MSZoning"] = df["MSZoning"].fillna(df["MSZoning"].mode()[0])

df["Exterior1st"] = df["Exterior1st"].fillna(df["Exterior1st"].mode()[0])

df["BsmtFinSF2"] = df["BsmtFinSF2"].fillna(df["BsmtFinSF2"].median())

df["TotalBsmtSF"] = df["TotalBsmtSF"].fillna(df["TotalBsmtSF"].median())

#Encoding

df["MSZoning"] = df["MSZoning"].astype("category").cat.codes

df["LotConfig"] = df["LotConfig"].astype("category").cat.codes

df["BldgType"] = df["BldgType"].astype("category").cat.codes

df["Exterior1st"] = df["Exterior1st"].astype("category").cat.codes


```

```python
# S - 2 X --> Y

X = df.drop(columns=["SalePrice"])

Y = df["SalePrice"]
```

```python
# S - 3 Train , test , split

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
# S - 4 Choose Model

from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, Y_train)
```

```python
# prediction with new data

new_house = pd.DataFrame({
    "Id":[2000],
    "MSSubClass":[60],
    "MSZoning":[3],
    "LotArea":[8500],
    "LotConfig":[1],
    "BldgType":[0],
    "OverallCond":[7],
    "YearBuilt":[1990],
    "YearRemodAdd":[2015],
    "Exterior1st":[5],
    "BsmtFinSF2":[0],
    "TotalBsmtSF":[900]
})
prediction = model.predict(new_house)

print("Predicted House Price =", prediction[0])
```

```python
from sklearn.metrics import r2_score
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error

import numpy as np
Y_pred = model.predict(X_test)

print("R2 Score :", r2_score(Y_test, Y_pred))

print("MAE :", mean_absolute_error(Y_test, Y_pred))

print("RMSE :", np.sqrt(mean_squared_error(Y_test, Y_pred)))
```
